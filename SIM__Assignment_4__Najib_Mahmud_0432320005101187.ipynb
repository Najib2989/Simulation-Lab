```python
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt
```

```python
Data_1 = [18.4, 19.1, 17.9, 18.7, 18.9]
Data_2 = [20.2, 20.5, 20.1, 20.3, 20.4]
```

```python
t_statistic, p_value = stats.ttest_ind(Data_1, Data_2, equal_var=True)
```

```python
n1, n2 = len(Data_1), len(Data_2)
s1_sq, s2_sq = np.var(Data_1, ddof=1), np.var(Data_2, ddof=1)
mean_diff = np.mean(Data_1) - np.mean(Data_2)
df = n1 + n2 - 2
```

```python
print(f"Sample 1 Mean: {np.mean(Data_1):.2f}")
print(f"Sample 2 Mean: {np.mean(Data_2):.2f}")
print(f"Sample Variance 1 : {s1_sq:.2f}")
print(f"Sample Variance 2 : {s2_sq:.2f}")
print(f"Mean Difference    : {mean_diff:.2f}\n")
print(f"t-Statistic: {t_statistic:.2f}")
print(f"Degrees of Freedom: {df:.2f}")
print(f"p-Value: {p_value:.4f}")
```

```python
sp_sq = ((n1 - 1) * s1_sq + (n2 - 1) * s2_sq) / (n1 + n2 - 2)
std_err = np.sqrt(sp_sq * (1/n1 + 1/n2))
ci = stats.t.interval(0.95, df, loc=mean_diff, scale=std_err)
```

```python
print(f"Confidence Interval of Mean Difference: [{ci[0]:.2f}, {ci[1]:.2f}]")
```

```python
alpha = 0.05
if p_value < alpha:
    print("Reject the null hypothesis. The means of the two samples are significantly different.")
else:
    print("Fail to reject the null hypothesis. The means of the two samples are not significantly different.")
```

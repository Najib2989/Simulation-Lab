```python
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt
```

```python
uniform_data = np.random.rand(10)
print("Uniform Random Numbers (1x10 Vector):")
print(uniform_data)

plt.figure(figsize=(6, 4))
plt.hist(uniform_data, bins=5, edgecolor='black', color='Blue')
plt.title('Histogram of Uniform Random Numbers (1x10)')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.show()
```

```python
normal_data = np.random.normal(loc=0, scale=1, size=500)

plt.figure(figsize=(6, 4))
plt.hist(normal_data, bins=30, edgecolor='black', color='Blue')
plt.title('Histogram of Normal Distribution (N=500)')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.show()
```

```python
print("Random Integers (3x3 Matrix):")

np.random.seed(10)
matrix_with_seed = np.random.randint(1, 51, size=(3, 3))
print("With fixed seed (np.random.seed(10)): ", matrix_with_seed)

np.random.seed(None)
matrix_without_seed = np.random.randint(1, 51, size=(3, 3))
print("Without setting a seed: ", matrix_without_seed)
```

```python
data = [5.2, 5.4, 5.1, 5.6, 5.3, 5.8]
hypothesized_mean = 5.5
t_stat, p_val = stats.ttest_1samp(data, hypothesized_mean)

print("One Sample T-Test Results: ")
print(f"T-statistic: {t_stat:.4f}")
print(f"P-value: {p_val:.4f}")
```

```python
alpha = 0.05
if p_val < alpha:
    print(f"Reject the null hypothesis. The average class score is significantly different from {hypothesized_mean:.2f}.")
else:
    print(f"Fail to reject the null hypothesis. The average class score is not significantly different from {hypothesized_mean:.2f}. ")
```

```python
plt.figure(figsize=(6, 5))
plt.bar(1, np.mean(data), color="skyblue", label='Sample Mean', width=0.4)
plt.errorbar(1, np.mean(data), yerr=np.std(data, ddof=1), fmt='k', linewidth=1.5, capsize=5)
plt.axhline(y=hypothesized_mean, color='r', linestyle='--', linewidth=1.5, label='Hypothesized Mean')
plt.title('Sample Mean vs Hypothesized Mean')
plt.xlim(0.5, 1.5)
plt.xlabel('Sample')
plt.ylabel('Mean Value')
plt.legend()
plt.grid(True)
plt.show()
```

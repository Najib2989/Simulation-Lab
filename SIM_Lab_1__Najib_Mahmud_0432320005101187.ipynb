```python
import numpy as np
from sympy import Matrix
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
scalar = 3.7
arr = np.array([1, 2, 3, 4, 5, 6])

sin_values = np.sin(arr)
cos_values = np.cos(arr)
tan_values = np.tan(arr)
asin_values = np.arcsin(np.clip(arr / np.max(arr), -1, 1))
acos_values = np.arccos(np.clip(arr / np.max(arr), -1, 1))
atan_values = np.arctan(arr)
abs_values = np.abs(arr)
sqrt_values = np.sqrt(arr)
floor_values = np.floor(arr)
ceil_values = np.ceil(arr)

print("sin:", sin_values)
print("cos:", cos_values)
print("tan:", tan_values)
print("asin:", asin_values)
print("acos:", acos_values)
print("atan", atan_values)
print("Absolute value:", abs_values)
print("Squer root:", sqrt_values)
print("floor:", floor_values)
print("ceil:", ceil_values)
```

```python
matrix = np.array([ [2.1, 4.7, 1.5], [6.8, 3.4, 5.2]])

max_value = np.max(matrix)
max_index = np.argmax(matrix)
min_value = np.min(matrix)
min_index = np.argmin(matrix)
matrix_length = len(matrix)
sum_matrix = np.sum(matrix)
prod_matrix = np.prod(matrix)
median_matrix = np.median(matrix)
mean_matrix = np.mean(matrix)
std_matrix = np.std(matrix)

print("maximum value:", max_value)
print("maximum index:", max_index)
print("minimum value:", min_value)
print("minimum index:", min_value)
print("vector length:", matrix_length)
print("Sum:", sum_matrix)
print("Product:", prod_matrix)
print("Median:", median_matrix)
print("Mean:", mean_matrix)
print("Standard Deviation:", std_matrix)
```
